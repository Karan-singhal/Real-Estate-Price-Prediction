```python
def convert_sqft(value):
    value=str(value).strip()
    #handle ranges  
    if "-" in value:
        parts=value.split("-")
        try:
            low=float(parts[0])
            high = float(parts[1])
            return (low+high)/2
        except:
            return None
    #sqaure meter
    if "Sq. Meter" in value:
        try:
            number = float(value.replace("Sq. Meter","").strip())
            return number * 10.7639
        except:
            return None

    #Square yard
    if "Sq. Yards" in value:
        try:
            number = float(value.replace("Sq. Yards", "").strip())
            return number * 9
        except:
            return None

    # Perch
    if "Perch" in value:
        try:
            number = float(value.replace("Perch", "").strip())
            return number * 272.25
        except:
            return None

    # Guntha
    if "Guntha" in value:
        try:
            number = float(value.replace("Guntha", "").strip())
            return number * 1089
        except:
            return None

    # Acre
    if "Acres" in value:
        try:
            number = float(value.replace("Acres", "").strip())
            return number * 43560
        except:
            return None

    # Cent
    if "Cents" in value:
        try:
            number = float(value.replace("Cents", "").strip())
            return number * 435.6
        except:
            return None

    # Ground
    if "Grounds" in value:
        try:
            number = float(value.replace("Grounds", "").strip())
            return number * 2400
        except:
            return None

    # Normal numeric value
    try:
        return float(value)
    except:
        return None
```

```python
df_clean["total_sqft_clean"] = df_clean["total_sqft"].apply(convert_sqft)
```
